bold text##  Step 1 — Imports & Configuration

```python
import random
import time

# ─── Difficulty Settings ───────────────────────────────────────────
DIFFICULTY = {
    'easy':   {'range': (1, 50),  'attempts': 10, 'label': '🟢 Easy'},
    'medium': {'range': (1, 100), 'attempts': 7,  'label': '🟡 Medium'},
    'hard':   {'range': (1, 200), 'attempts': 5,  'label': '🔴 Hard'},
}

print("✅ Libraries loaded successfully!")
print("⚙️  Difficulty levels configured:", list(DIFFICULTY.keys()))
```

##  Step 2 — Welcome Screen & Utilities

```python
def print_banner():
    """Display the welcome banner."""
    print("\n" + "═" * 55)
    print("  🎯  N U M B E R   G U E S S I N G   G A M E  🎯")
    print("═" * 55)
    print("  Can you crack the secret number?")
    print("  Rules: Guess the number with limited attempts.")
    print("         Hints: HOT 🔥 = very close | COLD ❄️ = far")
    print("═" * 55 + "\n")

def get_hint(secret: int, guess: int) -> str:
    """Return a proximity hint string."""
    diff = abs(secret - guess)
    direction = "📈 Go HIGHER" if guess < secret else "📉 Go LOWER"
    if diff == 0:
        return "🎉 CORRECT!"
    elif diff <= 5:
        return f"🔥 BOILING hot! {direction}"
    elif diff <= 15:
        return f"♨️  Getting warm... {direction}"
    elif diff <= 30:
        return f"🌤  Lukewarm. {direction}"
    else:
        return f"❄️  Ice cold! {direction}"

def display_progress_bar(attempts_left: int, max_attempts: int):
    """Visual attempts bar."""
    filled = attempts_left
    empty = max_attempts - attempts_left
    bar = '💚' * filled + '🖤' * empty
    print(f"  Attempts left: [{bar}] {attempts_left}/{max_attempts}")

print_banner()
print("✅ Utility functions ready!")
```

##  Step 3 — Core Game Logic

```python
def play_round(difficulty: str = 'medium') -> dict:
    """
    Play a single round of the guessing game.

    Parameters
    ----------
    difficulty : str
        One of 'easy', 'medium', 'hard'.

    Returns
    -------
    dict with keys: won, secret, guesses, time_taken, score
    """
    cfg = DIFFICULTY[difficulty]
    lo, hi = cfg['range']
    max_attempts = cfg['attempts']

    secret = random.randint(lo, hi)
    guesses = []
    start_time = time.time()

    print(f"\n{'─'*50}")
    print(f"  {cfg['label']} mode | Range: {lo}–{hi} | Attempts: {max_attempts}")
    print(f"{'─'*50}")
    print(f"  🤫 I'm thinking of a number between {lo} and {hi}...")
    print()

    for attempt in range(1, max_attempts + 1):
        display_progress_bar(max_attempts - attempt + 1, max_attempts)
        try:
            raw = input(f"  Attempt {attempt}: Enter your guess → ")
            guess = int(raw.strip())
        except ValueError:
            print("  ⚠️  Please enter a valid integer.")
            continue

        if guess < lo or guess > hi:
            print(f"  ⚠️  Number must be between {lo} and {hi}!")
            continue

        guesses.append(guess)
        hint = get_hint(secret, guess)
        print(f"  {hint}\n")

        if guess == secret:
            elapsed = round(time.time() - start_time, 1)
            score = max(100 - (attempt - 1) * 15 - int(elapsed), 10)
            print(f"  🏆 You got it in {attempt} attempt(s) and {elapsed}s!")
            print(f"  ⭐ Score: {score} points")
            return {'won': True, 'secret': secret, 'guesses': guesses,
                    'time_taken': elapsed, 'score': score}

    elapsed = round(time.time() - start_time, 1)
    print(f"  💀 Out of attempts! The number was {secret}.")
    return {'won': False, 'secret': secret, 'guesses': guesses,
            'time_taken': elapsed, 'score': 0}

print("✅ Game logic defined!")
```

##  Step 4 — Statistics Tracker

```python
def display_stats(history: list):
    """Print end-of-session statistics."""
    if not history:
        return
    total = len(history)
    wins  = sum(1 for r in history if r['won'])
    total_score = sum(r['score'] for r in history)
    avg_time    = sum(r['time_taken'] for r in history) / total
    avg_guesses = sum(len(r['guesses']) for r in history) / total

    print("\n" + "═" * 50)
    print("          📊  S E S S I O N   S T A T S")
    print("═" * 50)
    print(f"  Rounds played  : {total}")
    print(f"  Rounds won     : {wins}  ({wins/total*100:.0f}%)")
    print(f"  Total score    : {total_score} pts")
    print(f"  Avg guesses    : {avg_guesses:.1f}")
    print(f"  Avg time/round : {avg_time:.1f}s")
    print("═" * 50)
    if wins == total:
        print("  🎖  PERFECT SESSION — You're a legend!")
    elif wins >= total // 2:
        print("  👏  Good job! Keep practising.")
    else:
        print("  💪  Don't give up — try again!")
    print("═" * 50 + "\n")

print("✅ Stats tracker defined!")
```

##  Step 5 — Demo (Automated / No Input Required)

The cell below simulates a full game session automatically so the notebook can be run from top to bottom without requiring keyboard input.

```python
import random, time

def simulate_round(difficulty: str = 'medium', seed: int = 42) -> dict:
    """Automated demo — computer guesses using binary search."""
    random.seed(seed)
    cfg = DIFFICULTY[difficulty]
    lo_r, hi_r = cfg['range']
    max_attempts = cfg['attempts']

    secret = random.randint(lo_r, hi_r)
    lo_g, hi_g = lo_r, hi_r
    guesses = []
    start = time.time()

    print(f"\n  🤖 Auto-demo | {cfg['label']} mode")
    print(f"  Secret number (revealed for demo): {secret}")
    print(f"  {'─'*40}")

    for attempt in range(1, max_attempts + 1):
        guess = (lo_g + hi_g) // 2
        guesses.append(guess)
        hint = get_hint(secret, guess)
        print(f"  Attempt {attempt}: Guess = {guess:>4}  →  {hint}")
        if guess == secret:
            elapsed = round(time.time() - start, 3)
            score = max(100 - (attempt - 1) * 15, 10)
            print(f"  ✅ Found in {attempt} attempt(s)! Score: {score}")
            return {'won': True, 'secret': secret, 'guesses': guesses,
                    'time_taken': elapsed, 'score': score}
        if guess < secret:
            lo_g = guess + 1
        else:
            hi_g = guess - 1

    print(f"  ❌ Failed. Number was {secret}.")
    return {'won': False, 'secret': secret, 'guesses': guesses,
            'time_taken': round(time.time() - start, 3), 'score': 0}

print_banner()
history = []
for diff, seed in [('easy', 10), ('medium', 42), ('hard', 99)]:
    result = simulate_round(diff, seed)
    history.append(result)

display_stats(history)
```
