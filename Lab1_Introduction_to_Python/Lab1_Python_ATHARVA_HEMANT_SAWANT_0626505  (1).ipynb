## 1. Confusion Matrix Metrics (Python as a Calculator)
TP = 50
FP = 8
TN = 32
FN = 10

Total Predictions = TP + FP + TN + FP = 100 # Formaula - (a)
Accuracy = (TP + TN)/Total = 82/100 = 82% # Formula - (b)
Precision = TP/(TP+FP) = 50/(50+8) = 50/58 = 86.2% # Formula - (c)
Recall(Sensitivity) = TP/(TP+FN) = 50/(50+10) = 83.33% # Formula - (d)
F1 Score = 2*Precision*Recall/(Precision+Recall) = 2*0.862*0.833/(0.862+0.833) = 84.7% # Formula - (e)

## 2. Calculating R2 (Coefficient of Determinant)

. Sum of Squares due to Regression (SSR) = 420
. Sum of Squares due to Error (SSE) = 180

SST = SSR(420) + SSE(180) = 600
R2 = SSR/SST = 420/600 = 70%
